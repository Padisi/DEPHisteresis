```python
import numpy as np 
import matplotlib.pyplot as plt
```

# Pe > ROI

```python
k0 = 1
r_ROI = 1
r_Pe = 1.4
V_Pe = 4/3*np.pi*(r_Pe**3-r_ROI**3)
V_ROI = 4/3*np.pi*r_ROI**3
c_0 = 0.00001
N_Pe = V_Pe*c_0
N_ROI = V_ROI*c_0
N_Pe_history = [N_Pe]
N_ROI_history = [N_ROI]
max_frac = 0.64
dt = 0.01
for t in range(1000):
    k_ROI = N_Pe/dt*(1-N_ROI/(max_frac * V_ROI))
    k_Pe = k0*r_Pe**2*(1-N_Pe/(max_frac * V_Pe)) - k_ROI

    N_Pe += k_Pe*dt
    N_ROI += k_ROI*dt
    N_Pe_history.append(N_Pe)
    N_ROI_history.append(N_ROI)

k_out = 0.5
for t in range(4000):
    k_ROI = -k_out*r_ROI**2*N_ROI/(max_frac * V_ROI)
    k_Pe = -k_ROI - k_out*r_Pe**2*N_Pe/(max_frac * V_Pe)

    N_Pe += k_Pe*dt
    N_ROI += k_ROI*dt
    N_Pe_history.append(N_Pe)
    N_ROI_history.append(N_ROI)
    
for t in range(1000):
    k_ROI = N_Pe/dt*(1-N_ROI/(max_frac * V_ROI))
    k_Pe = k0*r_Pe**2*(1-N_Pe/(max_frac * V_Pe)) - k_ROI

    N_Pe += k_Pe*dt
    N_ROI += k_ROI*dt
    N_Pe_history.append(N_Pe)
    N_ROI_history.append(N_ROI)

plt.plot(N_Pe_history, label='N_Pe')
plt.plot(N_ROI_history, label='N_ROI')
plt.xlabel('Time step')
plt.ylabel('Number of particles')
plt.legend()
plt.show()
```
